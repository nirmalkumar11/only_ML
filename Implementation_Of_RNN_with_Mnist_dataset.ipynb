#**RNN**

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.layers import LSTM
```

```python
data=keras.datasets.mnist
```

```python
(image_a,label_a), (image_b, label_b)=data.load_data()
```

```python
image_a, image_b = image_a/255, image_b/255
```

```python
from tensorflow.keras.layers import Input
```

```python
image_a.shape
```

```python
input=Input(shape=(28,28))
rnn_layer=LSTM(128)(input)
output=layers.Dense(10, activation='softmax')(rnn_layer)
model=keras.Model(inputs=input, outputs=output)
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy']) #sparse-categorical-crossentrophy
```

```python
history=model.fit(image_a,label_a, validation_split=0.1,epochs=5, )
```

```python
import matplotlib.pyplot as plt
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.xlabel('epochs')
plt.ylabel('accuracy score')
plt.legend(['accuracy', 'val_accuracy'])
plt.show()
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.xlabel('epochs')
plt.ylabel('loss score')
plt.legend(['loss', 'val_loss'])
plt.show()
```

```python
loss,accurasy=model.evaluate(image_b,label_b)
```
