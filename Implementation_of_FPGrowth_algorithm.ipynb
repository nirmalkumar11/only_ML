#**FP Growth algorithm**

Which is similar to the Apriori algorithm, we use BFS Approach.

1. First we want to find the ione tem set of ites. choose the item set which have the support greater or equal to the minimum support.
2. order this frequent item set in decending order based on the support count.
3. find orderd item set, we order all items in the tranction based on it'd frequensy
4. with that orders item transaction, create a tree start with root node as none.
5. firsr create branch for first transaction.
6. for second transaction, continue with firsr transacton branch. if any item is present in first transacton branch that not present in the transaction, split the branch and add the next item in the new branch
7. create branch for all transaction.
8. create the conditional pattern base, which contains the path for each item in thet tree with the frequensy of thet item in that particular path.
9. conditional frequent pattern tree, which contains the frequent item for each item in conditional pattern base and sum of the frequensy.
10. then we find the frequent pattern, which is frequent item in conditional frequent pattern tree and the coresponding item. the confidence is sum of the frequensy in conditional frequent pattern tree.
11. maintain the rules, which confidence is greater or equal to the minimum confidence.

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import fpgrowth
from mlxtend.frequent_patterns import association_rules
```

```python
df=pd.read_csv('/content/drive/MyDrive/AI Training/Datasets/Copy of Market_Basket_Optimisation.csv', header=None)
```

```python
transaction=[]

for i in range(len(df)):
  transaction.append([i for i in df.iloc[i] if not pd.isna(i)])
```

```python
te = TransactionEncoder()
te_ary = te.fit_transform(transaction)
df_fp = pd.DataFrame(te_ary, columns=te.columns_)

display(df_fp.head())
```

```python
frequent_itemsets_fp = fpgrowth(df_fp, min_support=0.01, use_colnames=True)

print("Frequent Itemsets:")
display(frequent_itemsets_fp.head())
```

```python
rules_fp = association_rules(frequent_itemsets_fp, metric="confidence", min_threshold=0.2)

print("Association Rules:")
display(rules_fp.head())
```
