```python
from google.colab import userdata
from groq import Groq

groq_api_key = userdata.get('GROQ_API_KEY')

```

```python
sent= [
    "Transformer models use self-attention mechanisms to capture long-range dependencies in sequential data.",
    "Vector databases store high-dimensional embeddings to enable efficient similarity search for unstructured text.",
    "rag combines a retriever module with a language model to improve factual accuracy.",
    "Sentence embeddings are dense vector representations that encode semantic meaning of text inputs.",
    "FAISS is a library designed for fast nearest neighbor search in large-scale embedding spaces.",
    "Tokenization is the process of splitting raw text into smaller units such as words or subwords for model processing."
]
```

transformer model must be good at symantics search, multilingual

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
embedings=model.encode(sent).astype('float32')
```

```python
embedings.shape
```

```python
dim=embedings.shape[1]
```

```python
dim
```

```python
import faiss
index=faiss.IndexFlatL2(dim)
index.add(embedings)
```

```python
query='use of Transformer models'
```

```python
impyt_encode = model.encode([query]).astype('float32')
distances, indices = index.search(impyt_encode, 2)
```

```python
print(indices[0])
```

```python
con=''
for j in indices[0]:
    s=''
    for i in sent[j]:
      s+=i
    con+=s
    print(s)
```

```python
prompt=f"""
Answer the query based on the given context below.
context:{con}
query:{query}
answer formate: it must be polite and prefestional
"""
```

```python
client = Groq(api_key=groq_api_key)

completion = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[
        {"role": "user", "content": prompt}
    ],temperature=0.2
)

print(completion.choices[0].message.content)
```
