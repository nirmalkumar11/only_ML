```python
from sentence_transformers import SentenceTransformer
```

```python
model=SentenceTransformer('all-MiniLM-L6-v2')
```

```python
embedings=model.encode(resume).astype('float32')
```

```python
from google.colab import userdata
chroma_api=userdata.get('chroma_api')
```

```python
import chromadb

client = chromadb.CloudClient(
  api_key=chroma_api,
  tenant='91c18362-20dd-4ba0-a704-24d0151a6188',
  database='first_database'
)
```

```python
connection=client.get_or_create_collection(name='first_collection')
```

```python
connection.add(
    ids=[str(i) for i in range(len(resume))],
    embeddings=embedings
)
```

```python
prompt='what is the qualification of nirmalkumar'
context=connection.query(query_texts=prompt, n_results=1)
```

```python
!pip install groq
```

```python
prompt=f"""
response to the user query based on only the given context. dont generate your own story. the response only based on the given context.

context={context}

prompt={prompt}
"""
```

```python
groq_api_key = userdata.get('GROQ_API_KEY')
```

```python
from groq import Groq
client=Groq(api_key=groq_api_key)
```

```python
res=client.chat.completions.create(
    model='llama-3.1-8b-instant',
    messages=[{'role':'user', 'content':prompt}],
    temperature=0.1
)
```

```python
print(res.choices[0].message.content)
```
