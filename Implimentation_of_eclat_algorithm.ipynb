#**Eclat Algorithm**


    The Eclat algorithm is a frequent itemset mining technique.
    used in machine learning and data mining to discover patterns in transactional datasets.
    It is particularly useful for market basket analysis.
    where you want to find associations between items frequently bought together

**How Eclat Algorithm Works:**

The Eclat algorithm, short for “Equivalence Class Clustering and Bottom-Up Lattice Traversal”.

it is a depth-first search-based approach to find frequent itemsets.

It relies on the concept of an “equivalence class”(subset of large set) to reduce the search space efficiently.

because of DFS property, it becomes best method.

**Steps**


1. convert the horizontal dataset in to vertical dataset.
2. find first item set, and it's item count.
3. filter the T-id that have greater or equal element to the minimum support count.
4. take unique item from that first item set, and create two item set.
5. continue this process till we no more item set have been made.

    itemset-> frequent item set -> unuque T-id -> item set one increased count ->repeate
6. Find association rules, use multiple combinations.

   Ex:
   
   Item set={i1, i2, i3}

   Rules:

      1.i1->{i2,i3}

      2.i2->{i1,i3}
      
      3.13->{i1,i2}
7. Find confidence and select the rules with greater or equl to the minimum confidence.

**Formulas:**
$$
Support(X) = \frac{Count(TID)}{\text{Total Transactions}}
$$
$$
$$
$$
Support(X\rightarrow Y)=\frac{\text{Transactions}(X\cup Y)}{\text{Total Transactions}}
$$
$$
$$
$$
Confidence(X\rightarrow Y)=\frac{Support(X\cup Y)}{Support(X)}
$$
$$
$$
$$
Lift(X\rightarrow Y)=\frac{Confidence(X\rightarrow Y)}{Support(Y)}
$$
Lift > 1: Positive association.

Lift = 1: No association.

Lift < 1: Negative association.

*Note*: When we work on three or more itemset, we want to chek it is also a two item frequent set.

```python
eclatdata={}

for i in unique_values:
  ids=[]
  for k in range(len(eclatdf)):
    if i in list(eclatdf.iloc[k].values):
      ids.append(k)

  eclatdata[i]=set(ids)
```

```python
print(len(eclatdata))
```

```python

min_support = 2


frequent_itemsets = []

def eclat(prefix, items):

    while items:

        item, tidset = items.pop()

        support = len(tidset)

        if support >= min_support:

            new_itemset = prefix + [item]

            frequent_itemsets.append((new_itemset, support))

            new_items = []

            for other_item, other_tidset in items:

                intersection = tidset & other_tidset

                if len(intersection) >= min_support:

                    new_items.append(
                        (other_item, intersection)
                    )

            eclat(new_itemset, new_items)

items = list(eclatdata.items())

eclat([],  items)

print("\nFrequent Itemsets:\n")

for itemset, support in frequent_itemsets:

    print(itemset, "Support =", support)
```
