```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
y.head()
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test=train_test_split(x, y, random_state=42, test_size=0.2)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
trainsc=StandardScaler()
testsc=StandardScaler()
for i in x_train.columns.to_list():
  x_train[i]=trainsc.fit_transform(x_train[[i]])
  x_test[i]=testsc.fit_transform(x_test[[i]])
```

```python
import tensorflow as tf
from tensorflow import keras
```

tensore stores all data in multidimentional array, it is called as tensore,

<img src="https://tensorflownet.readthedocs.io/en/latest/_static/tensor-naming.png" width=500/>

we want to convert the nd input in to 1d array, we use flatten function to convert it into 1d array. it is used in input layer.

then we want to create hiddn\en layer, we use .dense to create hidden layer.

for output layer also, we use .dense with sigmoid as activation.



```python
model=keras.Sequential([
    keras.layers.Flatten(input_shape=(30,)),
    keras.layers.Dense(20, activation='relu'),
    keras.layers.Dense(2, activation='sigmoid')
])
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

How we give the correct number of nodes for layers.



```python
histry=model.fit(x_train, y_train, validation_split=0.1, epochs=60)
```

```python
plt.plot(histry.history['accuracy'])
plt.plot(histry.history['val_accuracy'])

plt.title("accuracy vs val_accuracy")
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend(['training data', 'validation data'], loc ='lower right')
```

```python
plt.plot(histry.history['loss'])
plt.plot(histry.history['val_loss'])
plt.title('val_loss ns loss')
plt.xlabel('epoch')
plt.ylabel('loss')
plt.legend(['training data', 'testing data'], loc='lower right')
```
